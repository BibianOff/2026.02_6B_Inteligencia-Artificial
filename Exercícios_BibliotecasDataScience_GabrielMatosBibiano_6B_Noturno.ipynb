```python
#19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

vetor1 = np.random.randint(1, 50, size=20)
vetor2 = np.random.randint(1, 50, size=20)

produto_escalar = np.dot(vetor1, vetor2)
norma_v1 = np.linalg.norm(vetor1)
norma_v2 = np.linalg.norm(vetor2)

cosseno_theta = produto_escalar / (norma_v1 * norma_v2)
cosseno_theta = np.clip(cosseno_theta, -1.0, 1.0)

angulo_rad = np.arccos(cosseno_theta)
angulo_graus = np.degrees(angulo_rad)

print(f"Produto Escalar: {produto_escalar}")
print(f"Cosseno do ângulo: {cosseno_theta:.4f}")
print(f"Ângulo entre os vetores: {angulo_graus:.2f} graus")
```

```python
#1. Calcule a integral da função f(x) = x^2 sin(x) no intervalo de 0 a pi.

from scipy.integrate import quad

def f(x):
  return (x**2) * np.sin(x)

a = np.pi
result, erro = quad(f,0,a)
print (f"Resultado da integral: {(result): .2f}")
print (f"Erro: {(erro)}")
```

```python
#2. Resolva a equação diferencial dy/dx = x - y, com y(0) = 1,
#no intervalo de 0 a 10.

from scipy.integrate import solve_ivp

def eq_dif(x,y):
  return x - y
result = solve_ivp(eq_dif, [0,10], [1])
print("Tempo (ou X):", result.t)
print("Solução para Y:", result.y[0])
```

```python
#3. Encontre as raízes da função f(x) = x^4 - 3x^3 + 2 utilizando SciPy.

from scipy.optimize import root

def f(x):
  return (x**4) - (3*(x**3)) + 2

solucao = root (f,1)
print (f"Raizes:\n {(solucao)}")
```

```python
#4. Use a função de otimização de SciPy para maximizar f(x) = -x^2 + 4x + 1

from scipy.optimize import minimize

def f(x):
  return ((-x**2) + (4*x) + 1) * -1

result = minimize (f , 0)
max_valor = -result.fun
max_ponto = result.x
print(f"Máximo: {max_valor} em x = {max_ponto}")


```

```python
#5. Calcule a transformada de Fourier de um sinal composto de
#sin(2*pi*t) e cos(4*pi*t).

from scipy.fft import fft

t = np.linspace(0, 1, 500)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
frenqencia = fft(sinal)
print("Transformada de Fourier:", frenqencia)
```

```python
#6. Encontre a matriz inversa de uma matriz 5x5 de números aleatórios.

from scipy.linalg import inv

matriz_exemplo = np.random.rand(5, 5)
matriz_inversa = inv(matriz_exemplo)
print("Matriz:\n", matriz_inversa)
```

```python
#7. Use a função de interpolação cúbica de SciPy para interpolar os dados
#(0, 0), (1, 2), (2, 3), (3, 5), (4, 4).

from scipy.interpolate import interp1d

x = np.array([0,1,2,3,4])
y = np.array([0,2,3,5,4])
func = interp1d(x,y,kind = 'cubic')
x1 = 2
y1 = func(x1)
print(f"O valor interpolado para x = {(x1)} é: {y1}")
```

```python
#8. Resolva o sistema de equações $3x + 2y = 5$ e $x + 4y = 6$.

from scipy import linalg

A = np.array([[3,2], [1,4]])
B = np.array([5,6])

result = linalg.solve(A,B)
print(f"Resulado de [A, B] é {(result)}")
```

```python
#9. Ajuste uma curva quadrática aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8).

from scipy.optimize import curve_fit

A = np.array([1,2,3,4,5])
B = np.array([2,4,5,7,8])

def y(x,a,b,c):
  return a * (x**2) + b * x + c
parametros, cov = curve_fit(y, A, B)
print("Parâmetros ajustados [a, b,c]:", parametros)
```

```python
#10. Calcule a integral dupla da função f(x, y) = xy + x^2 no intervalo
#[0, 2] para X e [0, 1] para Y.

from scipy.integrate import dblquad

def xy(y,x):
  return x*y + (x**2)

x_i, x_f = 0,2
y_i = lambda x: 0
y_f = lambda x: 1

resultado, erro = dblquad(xy, x_i, x_f, y_i, y_f)
print (" Resultado da integral :", resultado )
```

```python
#11. Encontre os autovalores e autovetores de uma matriz 4x4 de números inteiros
#aleatórios.

from scipy.linalg import eig

x = np.random.randint(1, 10, size=(4, 4))
autovalores, autovetores = eig(x)

print("Matriz Original:\n", x)
print("\nAutovalores:\n", autovalores)
print("\nAutovetores:\n", autovetores)
```

```python
#14. Utilize a função de otimização multivariada para minimizar
#f(x, y) = (x - 2)^2 + (y - 3)^2.

from scipy.optimize import minimize

def f(variaveis):
  x = variaveis[0]
  y = variaveis[1]

  return ((x-2)**2) + ((y-3)**2)

chute = [1, 1]
result = minimize (f , chute)
print(f"O ponto mínimo exato é [x, y]: {result.x}")


```

```python
#10.Plote três gráficos de linha em um único gráfico, cada um com uma cor e
#estilo de linha diferentes.

x = np.linspace(0.1, 10 , 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) + np.cos(x)
plt.plot(x,y1, color="Blue", linestyle='-', label = 'Sin')
plt.plot(x,y2, color="Green", linestyle='--', label = 'Cos')
plt.plot(x,y3, color="Purple", linestyle=':', label = 'Sin + Cos')
plt.legend()
plt.show()

```

```python
#18.Crie um gráfico de barras com valores positivos e negativos e use diferentes
#cores para cada tipo de valor.

x = ['A', 'B', 'C', 'D']
y = np.random.randint(-10, 10, size = 4)
cores = ['blue' if valor >= 0 else 'red' for valor in y]
plt.bar(x,y,color=cores)
plt.axhline(0, color='black', linewidth=1)
plt.title ("Grafico de Barras")
plt.xlabel("Valor")
plt.ylabel("Categoria")
plt.show()
```
