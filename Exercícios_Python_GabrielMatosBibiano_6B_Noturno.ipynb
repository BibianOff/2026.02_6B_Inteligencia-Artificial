```python
#21. Busca Binária

#O código transpõe uma matriz 3X3 convertendo colunas em linhas através
#do operador de desempacotamento (*) e da função zip, exibindo cada linha
#resultante via laço for.

def busca_binaria(lista, alvo):
    i, f = 0, len(lista) - 1
    while i <= f:
        m = (i + f) // 2
        if lista[m] == alvo:
            return m
        i, f = (m + 1, f) if lista[m] < alvo else (i, m - 1)
    return -1


nums = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(busca_binaria(nums, 5))

```

```python
#22. Cifra de Cesar

#O codigo cria uma Cifra de César deslocando cada letra do texto por
#um valor inteiro de y no alfabeto minúsculo importado pela biblioetca String,
#e do operador % para tratar a rotação cíclica e preservando caracteres não
#alfabéticos.

import string

x = input('Escreve algo ai: ').lower()
y = int(input('Digita Num: '))
senha = ''
base = string.ascii_lowercase
for i in x:
    if i in base:
        Base1 = base.find(i)
        Base2 = (Base1 + y) % len(base)
        senha += base[Base2]
    else:
        senha += i

print(f'Texto Cifrado: {senha}')
```

```python
#24 Validar CPF

#O sistema valida um CPF filtrando ele (str.isdigit) para isolar 11 dígitos
#e autenticando os dois dígitos verificadores por um do cálculo disponibilizado
#pelo Gov

def validar_cpf(cpf: str) -> bool:
  cpf = ''.join(filter(str.isdigit, cpf))
  if len(cpf) != 11:
      print("CPF inválido")


  soma = sum(int(cpf[i]) * (10 - i) for i in range(9))
  somaR = (soma * 10) % 11
  digito1 = 0 if somaR > 9 else somaR
  if digito1 != int(cpf[9]):
      print("CPF inválido")
      return False

  soma = sum(int(cpf[i]) * (11 - i) for i in range(10))
  somaR = (soma * 10) % 11
  digito2 = 0 if somaR > 10 else somaR
  if digito2 != int(cpf[10]):
      print("CPF inválido")
      if digito2 != int(cpf[10]):
          print("CPF inválido")
          return False


  return print("CPF válido")

cpf = input("Digite um CPF: ")
validar_cpf(cpf)


```

```python
#30 Jogo da forca

#O Sistema sorteia uma palavra via random.choice e gerenciando o estado da
#partida em um laço while, onde valida as entradas do usuário, rastreia acertos
#e erros em listas e verifica a condição de vitória por comparação de
#conjuntos (set).

import random

palavras = ['jogo', 'manga', 'mimikyu', 'teclado', 'monster']
palavra = random.choice(palavras)
letras_erradas = []
letras_certas = []
tentativas = 6

print("Jogo da força - 6 tentaivas \n")

while tentativas > 0 :
  progresso = [letra if letra in letras_certas else '_' for letra in palavra]
  print("Palavra:", " ".join(progresso))
  x = input("Digita uma letra: \n")

  if not x.isalpha() or len(x) != 1:
    print("Apenas uma LETRA\n")
    continue

  if x in letras_certas or x in letras_erradas:
    print(f"Letra já utilizada. Erradas: {letras_erradas} | Certas: {letras_certas}\n")
    continue

  if x in palavra:
    letras_certas.append(x)
    print("Acertou!\n")
  else:
    letras_erradas.append(x)
    tentativas -= 1
    print(f"Errou! Tentativas restantes: {tentativas}\n")


  if set(palavra) == set(letras_certas):
    print(f"Você ganhou! A palavra era: {palavra}")
    break
else:
    print(f"Você perdeu! A palavra secreta era: {palavra}")





```

```python
#31.Ordenação por Bolha

#O código sorteia 10 números aleatórios sem repetir e organiza em ordem
#crescente usando o Bubble Sort, trocando os números vizinhos de lugar e
#parando mais cedo se a lista já estiver pronta.

import random
numeros = random.sample(range(1, 100), 10)
n = len(numeros)

for i in range(n):
    trocou = False
    for j in range(0, n - i - 1):
        if numeros[j] > numeros[j + 1]:
            numeros[j], numeros[j + 1] = numeros[j + 1], numeros[j]
            trocou = True
    if not trocou:
        break

print(numeros)
```

```python
#32.Jogo da Velha

#O código implementa um jogo da velha interativo para terminal usando uma matriz
#3X3, gerenciando a alternância de turnos entre os jogadores ('X' e 'O'),
#tratando entradas inválidas com try/except e checando condições de vitória ou
#empate após cada movimento.

def mostrar_tabuleiro(tabuleiro):
    for linha in tabuleiro:
        print(" | ".join(linha))
        print("-" * 9)

def verificar_vitoria(tabuleiro, jogador):
    for linha in tabuleiro:
        if linha.count(jogador) == 3:
            return True

    for c in range(3):
        if tabuleiro[0][c] == tabuleiro[1][c] == tabuleiro[2][c] == jogador:
            return True

    if tabuleiro[0][0] == tabuleiro[1][1] == tabuleiro[2][2] == jogador:
        return True
    if tabuleiro[0][2] == tabuleiro[1][1] == tabuleiro[2][0] == jogador:
        return True

    return False

def verificar_empate(tabuleiro):
    for linha in tabuleiro:
        if " " in linha:
            return False
    return True

tabuleiro = [[" " for _ in range(3)] for _ in range(3)]
jogador_atual = "X"

while True:
    mostrar_tabuleiro(tabuleiro)
    print(f"\nVez do jogador {jogador_atual}")

    try:
        linha = int(input("Escolha a linha (0, 1 ou 2): "))
        coluna = int(input("Escolha a coluna (0, 1 ou 2): "))
    except ValueError:
        print("Digite apenas números inteiros (0, 1 ou 2)!\n")
        continue

    if linha not in [0, 1, 2] or coluna not in [0, 1, 2]:
        print("Posição fora do tabuleiro! Escolha valores entre 0 e 2.\n")
        continue

    if tabuleiro[linha][coluna] != " ":
        print("Essa posição já está ocupada! Tente outra.\n")
        continue

    tabuleiro[linha][coluna] = jogador_atual

    if verificar_vitoria(tabuleiro, jogador_atual):
        mostrar_tabuleiro(tabuleiro)
        print(f"\nParabéns! Jogador '{jogador_atual}' venceu!")
        break

    if verificar_empate(tabuleiro):
        mostrar_tabuleiro(tabuleiro)
        print("\nDeu velha! Empate!")
        break

    jogador_atual = "O" if jogador_atual == "X" else "X"
```

```python
#37. Funções Recursivas

#O código calcula o fatorial de um número inteiro de forma recursiva,
#tratando entradas negativas, definindo os casos base para 0 e 1 e multiplicando
#n pelo retorno de chamadas sucessivas a fatorial(n - 1).

def fatorial(n: int) -> int:
    if n < 0:
        print("Fatorial não é definido para números negativos.")
    if n == 0 or n == 1:
        return 1
    return n * fatorial(n - 1)
print(fatorial(5))
```

```python
#45.Par ou Ímpar em Lista

#O código gera 10 números aleatórios entre 1 e 100 via random.choices, separa
#os valores em listas de pares e ímpares com um laço for e o operador %
#e exibe ambos os conjuntos no final.

import random
Lista = random.choices(range(1, 101), k=10)
Par = []
Impar = []
for j in Lista:
  if j % 2 == 0:
    Par.append(j)
  else:
    Impar.append(j)
print(f"Lista Par:\n{(Par)}\n")
print(f"Lista Impar:\n{(Impar)}\n")
```

```python
#46. Calcular Média e Desvio Padrão

#O código gera uma lista com $x$ números aleatórios via random.choices,
#calcula a média aritmética e os desvios padrão amostral e populacional
#utilizando a soma dos quadrados das diferenças

import random
x = int(input('Digita Num: '))
Lista = random.choices(range(1, 101), k=x)
n = len(Lista)

media = sum(Lista) / n

somaQ = sum((x - media) ** 2 for x in Lista)

desvioP_amostral = (somaQ / (n - 1)) ** 0.5
desvioP_populacional = (somaQ / n) ** 0.5

print(f"Lista: {Lista}")
print(f"Média: {media:.2f}")
print(f"Desvio Padrão Amostral: {desvioP_amostral:.2f}")
print(f"Desvio Padrão Populacional: {desvioP_populacional:.2f}")
```

```python
#47. Contar Números em Lista

#O código gera uma lista de 20 números aleatórios entre 1 e 9 via random.choices
#elimina duplicatas e ordena os valores únicos com sorted(set()), exibindo a
#frequência de cada elemento através do método count().

import random

lista = random.choices(range(1, 10), k=20)
print(lista)
for x in sorted(set(lista)):
    print(f"O número {x} aparece {lista.count(x)} vezes")
```

```python
#48.Ordenação por Inserção

#O sistema gera 20 números aleatórios e os ordena em ordem crescente aplicando
#o Insertion Sort, que percorre a lista a partir do segundo elemento e desloca
#os itens maiores para a direita até encontrar a posição correta da chave atual.

import random
lista = random.choices(range(1, 100), k=20)

def insertion_sort(lista):
    for i in range(1, len(lista)):
        chave = lista[i]
        j = i - 1
        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1
        lista[j + 1] = chave

    return lista

print("Lista original:", lista)
insertion_sort(lista)
print("Lista ordenada:", lista)
```

```python
#49.Soma dos Dígitos

#O código recebe uma entrada, filtra apenas os caracteres numéricos com
#isdigit() e calcula a soma total de todos os dígitos utilizando a função sum().

numero = input("Digite Num: ")
soma = sum(int(digito)
  for digito in numero
  if digito.isdigit())

print(f"Soma dos dígitos: {soma}")
```

```python
#50. Contar Palavras em Texto

#O código converte o texto para minúsculas, isola e ordena as letras únicas com
#sorted(set()) e exibe a frequência de cada caractere alfabético através do
#método count().

x = input("Digita algo ai: ").lower()
for letra in sorted(set(x)):
    if letra.isalpha():
        print(f"A letra '{letra}' aparece {x.count(letra)} vezes")
```
